```python
import pandas as pd
import numpy as np

def setup_data():
    df = pd.read_csv("dataset.csv")
    id_vars = ['datetime', 'underlying_price']
    value_vars = [col for col in df.columns if col not in id_vars]

    df_long = pd.melt(df, id_vars=id_vars, value_vars=value_vars, 
                      var_name='ticker', value_name='implied_volatility')

    # Re-added the expiry_str parsing line
    df_long['option_type'] = df_long['ticker'].str[-2:]
    df_long['expiry_str'] = df_long['ticker'].str[5:12]
    df_long['strike'] = df_long['ticker'].str.extract(r'(\d+)(?:CE|PE)')[0].astype(float)
    df_long['moneyness'] = df_long['strike'] / df_long['underlying_price']
    df_long['parsed_dt'] = pd.to_datetime(df_long['datetime'], format='mixed', dayfirst=True)

    train = df_long[df_long['implied_volatility'].notna()].copy()
    test = df_long[df_long['implied_volatility'].isna()].copy()

    train.to_csv("train_dataset.csv", index=False)
    test.to_csv("test_dataset.csv", index=False)
    print("System setup complete: Train and Test datasets generated.")

setup_data()
```
